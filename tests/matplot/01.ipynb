```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig, ax = plt.subplots()
aa = ax.plot([1,2,3,4], [1,4,2,3])
plt.show()
print(aa)
```

```python
plt.hlines(y=[0.2, 0.4, 0.6], xmin=0, xmax=1, colors='purple', linestyles='dotted')
plt.vlines(x=[0.3, 0.7], ymin=0, ymax=1, colors='#FF8800')
plt.show()
```

```python
fig = plt.figure(figsize=(2, 2), facecolor='lightskyblue',
                 layout='constrained')
fig.suptitle('Figure')
ax = fig.add_subplot()
ax.set_title('Axes', loc='left', fontstyle='oblique', fontsize='medium')

bx = fig.add_subplot(pos="right")
bx.set_title('Axes', loc='left', fontstyle='oblique', fontsize='medium')
```
